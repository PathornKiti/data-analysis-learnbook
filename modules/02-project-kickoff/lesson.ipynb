# Module 2: Project kickoff, from interest to question

**Goal:** turn something you care about into a question that data can answer, and choose a dataset to answer it.

There is almost no code here on purpose. *Choosing a good question is the most important skill in data analysis.*

**Maths in this module:** none. Just careful thinking about what "answerable" means.

## 1. Four kinds of questions

| type | asks | example |
|---|---|---|
| **Describe** | What does it look like? | *What is the typical number of daily rides?* |
| **Compare** | Are groups different? | *Are weekends busier than weekdays?* |
| **Relate** | Do two things move together? | *Do rides rise when it gets warmer?* |
| **Predict** | Can we forecast something? | *How many rides tomorrow, given the forecast?* |

Descriptive and comparative questions come first; relating and predicting build on them.
Most good projects contain all four, in that order.

## 2. From vague interest to answerable question

Start with an interest, then sharpen it step by step.

| vague | better | answerable with data |
|---|---|---|
| "I like football" | "What makes teams win?" | "Is the number of shots on target related to goals scored per match, in the 2022–23 league season?" |
| "Climate is important" | "Is it getting hotter?" | "How has the yearly average temperature in my city changed since 1980?" |
| "I want to understand spending" | "Who spends more?" | "Do weekend orders differ in size from weekday orders?" |

A good question is **specific** (who/what/when), **measurable** (a column could answer it) and
**interesting to you** (you'll spend weeks with it).

```python
import pandas as pd

rides = pd.read_csv(DATA / "daily_rides.csv")
rides.head()
```

The **feasibility check** every project needs. Ask these of your data *before* committing:

1. **Does a column exist for each thing in my question?** (`rides`, `temp_c`, `rain_mm`: yes)
2. **Is there enough data?** (rows = days)
3. **Is it clean enough?** (some missing temperatures: manageable)
4. **Am I allowed to use it?** (check the licence; here the data is synthetic)

```python
print("rows, columns:", rides.shape)
print("columns:", list(rides.columns))
print("missing per column:")
print(rides.isna().sum())
```
