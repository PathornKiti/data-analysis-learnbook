## 1. Variables and types

A **variable** is a name for a value. Python values have **types**, and the type decides what you can do with it.

```python
age = 21               # int      (whole number)
height_m = 1.72        # float    (decimal number)
name = "Ada"           # str      (text)
is_student = True      # bool     (True / False)

print(name, age, height_m, is_student)
print(type(age), type(height_m), type(name), type(is_student))
```

## 2. Arithmetic (and the maths behind it)

| operator | meaning | example | result |
|---|---|---|---|
| `+  -  *  /` | add, subtract, multiply, divide | `7 / 2` | `3.5` |
| `**` | power (exponent) | `2 ** 3` | `8` |
| `//` | whole-number division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |

Python follows the same **order of operations** as maths (brackets → powers → ×÷ → +−).

```python
print(2 + 3 * 4)      # multiplication first
print((2 + 3) * 4)    # brackets first
print(2 ** 3 ** 2)    # powers group right-to-left: 2 ** 9
print(17 % 5)         # 17 = 3*5 + 2, so the remainder is 2
print(7 / 2, 7 // 2)
```

**Percentages and ratios** show up constantly in data work: *x percent of y* is `x / 100 * y`,
and *part as a percentage of whole* is `part / whole * 100`.

```python
passed, total = 38, 50
print(f"{passed / total * 100:.1f}% of students passed")   # an f-string: more below
```

## 3. Strings

Text is made of characters. **f-strings** (`f"..."`) let you drop values into text with `{}`.

```python
city = "bangkok"
print(city.upper(), len(city), city[0], city[-1])
print(city.title().replace("k", "K"))
print(f"{city.title()} has {len(city)} letters")
print("a,b,c".split(","))
```

## 4. Lists: ordered collections

A **list** holds many values in order. Counting starts at **0**.

```python
scores = [72, 85, 90, 64, 78]

print(scores[0], scores[-1])        # first and last
print(scores[1:3])                  # slice: positions 1 and 2 (end is excluded)
scores.append(95)                   # add to the end
print(len(scores), min(scores), max(scores), sum(scores))
print(sorted(scores))
```

## 5. Dictionaries: labelled values

A **dictionary** maps a *key* to a *value*, like a mini lookup table. Records of data often look like this.

```python
student = {"name": "Ada", "age": 21, "scores": [72, 85, 90]}

print(student["name"])
student["major"] = "Science"        # add a new key
print(student.keys())
print(student.get("height", "unknown"))   # safe lookup with a default
```

## 6. Making decisions: `if`

Indentation (4 spaces) is part of Python's syntax: it shows what belongs inside the `if`.

```python
def grade(score):
    if score >= 80:
        return "A"
    elif score >= 65:
        return "B"
    else:
        return "C"

print(grade(90), grade(70), grade(40))
print(5 > 3 and 2 > 4, 5 > 3 or 2 > 4, not (5 > 3))
```

## 7. Repeating things: loops

Use a `for` loop to do something for every item.

```python
for s in [72, 85, 90]:
    print(s, "->", grade(s))

for i in range(3):          # range(3) gives 0, 1, 2
    print("round", i)
```

### 🧮 Maths link: summation notation Σ **is** a loop

Mathematicians write the total of $x_1, x_2, \dots, x_n$ as

$$\sum_{i=1}^{n} x_i = x_1 + x_2 + \dots + x_n$$

and the **mean** (average) as $\bar{x} = \dfrac{1}{n}\sum_{i=1}^{n} x_i$.
In Python the Σ is just a loop that accumulates a total:

```python
x = [72, 85, 90, 64, 78]

total = 0
for value in x:
    total = total + value        # the Σ

mean = total / len(x)            # divide by n
print(total, mean)
print(sum(x) / len(x))           # the built-in shortcut gives the same answer
```

## 8. Functions

A **function** packages a recipe so you can reuse it. This is the same idea as a maths function
$f(x) = 2x + 1$: put a number in, get a number out.

```python
def f(x):
    return 2 * x + 1

print(f(0), f(1), f(10))

def mean(values):
    """Return the average of a list of numbers."""
    return sum(values) / len(values)

print(mean([72, 85, 90, 64, 78]))
```

> ✍️ The text in triple quotes right after `def` is a **docstring**: a one-line explanation of what the
> function does. Writing them is a habit worth forming now. You'll need it when your project code lives in `.py` files.

Functions can have default values and several inputs:

```python
def percent(part, whole, digits=1):
    return round(part / whole * 100, digits)

print(percent(38, 50))
print(percent(1, 3, digits=3))
```

## 9. List comprehensions: build lists in one line

`[expression for item in items if condition]` is a compact loop. You'll see it everywhere in data code.

```python
scores = [72, 85, 90, 64, 78]

doubled = [s * 2 for s in scores]
passing = [s for s in scores if s >= 70]
labels = [grade(s) for s in scores]

print(doubled)
print(passing)
print(labels)
```

## 10. Counting with a dictionary

A classic pattern: count how often each value appears.

```python
votes = ["tea", "coffee", "tea", "juice", "tea", "coffee"]

counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1

print(counts)
print(max(counts, key=counts.get))     # which key has the biggest count?
```

## 11. Reading errors

Errors are not failures. They are Python telling you what to fix. Read the **last line** first.

```python
try:
    print(int("twelve"))
except ValueError as err:
    print("Caught an error:", err)
```

| error | usual cause |
|---|---|
| `NameError` | typo in a name, or you didn't run the cell that defines it |
| `TypeError` | mixing types, e.g. `"5" + 5` |
| `IndexError` | list position that doesn't exist |
| `KeyError` | dictionary key that doesn't exist |
| `IndentationError` | inconsistent indentation |
