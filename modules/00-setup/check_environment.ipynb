```python
import importlib

needed = ["numpy", "pandas", "matplotlib", "seaborn", "scipy", "sklearn"]
missing = []
for name in needed:
    try:
        module = importlib.import_module(name)
        print(f"{name:12s} {getattr(module, '__version__', '?')}")
    except ImportError:
        missing.append(name)

assert not missing, f"Missing packages: {missing}. Run: pip install -r requirements.txt"
print("\n✅ Everything is installed.")
```

## A first taste of data analysis

Don't worry about understanding the code yet. By the end of the course you will be able to
write something like this yourself.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
hours = rng.uniform(0, 10, 50)
score = 40 + 5 * hours + rng.normal(0, 6, 50)

plt.scatter(hours, score)
plt.xlabel("Hours studied")
plt.ylabel("Score")
plt.title("Does studying help? (fake data)")
plt.show()
print("✅ Plotting works. Welcome aboard!")
```
