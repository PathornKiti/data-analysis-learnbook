# Module 3: NumPy, fast arrays of numbers

**Goal:** compute with whole collections of numbers at once, the foundation under pandas, scikit-learn and almost everything else.

**You will learn:** arrays · indexing and masks · vectorised maths · aggregation · broadcasting · random numbers

**Maths in this module:** vectors · dot product · norm (length) · matrices · z-scores (standardising)

```python
import numpy as np
```

## 1. Arrays

A NumPy **array** is like a list, but all elements share one type and maths applies to the whole thing at once.

```python
a = np.array([1, 2, 3, 4, 5])
print(a, a.dtype, a.shape)

print(np.arange(0, 10, 2))       # like range, but an array: 0 2 4 6 8
print(np.linspace(0, 1, 5))      # 5 evenly spaced numbers from 0 to 1
print(np.zeros(3), np.ones(3))
```

## 2. Vectorised maths: no loops needed

```python
celsius = np.array([0, 20, 37, 100])
fahrenheit = 9 / 5 * celsius + 32      # applied to every element
print(fahrenheit)

print(celsius ** 2)
print(np.sqrt(celsius))
```

Why bother? Speed. Compare adding up the squares of a million numbers with a loop vs. NumPy:

```python
import time

big = np.arange(1_000_000)

t0 = time.perf_counter()
total_loop = 0
for v in big.tolist():
    total_loop += v * v
t_loop = time.perf_counter() - t0

t0 = time.perf_counter()
total_np = (big * big).sum()
t_np = time.perf_counter() - t0

print(f"loop: {t_loop:.3f}s   numpy: {t_np:.4f}s   (about {t_loop / t_np:.0f}x faster)")
```

## 3. Indexing, slicing and **masks**

```python
temps = np.array([28.5, 31.2, 35.0, 29.9, 30.0, 33.4])

print(temps[0], temps[-1], temps[1:4])

mask = temps > 30                # an array of True/False
print(mask)
print(temps[mask])               # keep only where True
print((temps > 30).sum(), "days above 30")      # True counts as 1
print(temps[(temps > 29) & (temps < 33)])        # combine with & (and), | (or)
```

## 4. Two dimensions: rows and columns

A 2-D array is a table of numbers (a **matrix**). `shape` is `(rows, columns)`.

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)
print(m[0, 1], m[:, 2], m[1, :])    # element, whole column, whole row

print(m.sum(), m.sum(axis=0), m.sum(axis=1))   # all, per column, per row
print(m.mean(axis=0))
```

**`axis`** is the thing people trip over: `axis=0` collapses the **rows** (giving one value per column);
`axis=1` collapses the columns (one value per row).

## 5. Summary statistics

```python
scores = np.array([72, 85, 90, 64, 78, 95, 55])
print("mean  ", scores.mean())
print("median", np.median(scores))
print("std   ", scores.std().round(2))
print("min/max", scores.min(), scores.max())
print("75th percentile", np.percentile(scores, 75))
print("argmax (position of the max)", scores.argmax())
```

## 6. Random numbers (for simulation)

Use `default_rng(seed)`: the seed makes results repeatable, so you and a friend see the same numbers.

```python
rng = np.random.default_rng(seed=7)
print(rng.integers(1, 7, size=10))             # ten dice rolls
print(rng.normal(loc=50, scale=10, size=5))    # bell-curve numbers
print(rng.choice(["heads", "tails"], size=5))
```

## 🧮 7. Maths link: vectors, dot products and length

A **vector** is a list of numbers, and a NumPy 1-D array is exactly that. Think of a student as a vector
`[study_hours, sleep_hours]`. Vectors can be added and scaled just like arrays:

```python
u = np.array([3, 4])
v = np.array([1, 2])
print(u + v, 2 * u)
```

The **length** (norm) of a vector, via Pythagoras, is $\lVert u \rVert = \sqrt{u_1^2 + u_2^2 + \dots}$.
The **distance** between two points is the length of their difference. That is the idea behind
"find the most similar student".

```python
print(np.linalg.norm(u))         # sqrt(3² + 4²) = 5
print(np.linalg.norm(u - v))     # distance between u and v
```

The **dot product** multiplies matching entries and adds them up:

$$u \cdot v = u_1 v_1 + u_2 v_2 + \dots = \sum_i u_i v_i$$

It is the engine of regression: a prediction like `score = 2*study + 3*sleep` is *a dot product of
the weights `[2, 3]` with a student's vector*.

```python
weights = np.array([2.0, 3.0])
student = np.array([10, 7])          # 10 study hours, 7 sleep hours
print(np.dot(weights, student))      # 2*10 + 3*7 = 41
print(weights @ student)             # @ is the shortcut
```

A **matrix** stacks many vectors; `matrix @ vector` computes the dot product for every row at once.
Predicting for *many students in one line* is the payoff.

```python
students = np.array([[10, 7],
                     [ 2, 5],
                     [ 8, 8]])
print(students @ weights)          # one prediction per student
```

## 🧮 8. Maths link: standardising with z-scores

How unusual is a value? Subtract the mean and divide by the standard deviation:

$$z = \frac{x - \bar{x}}{s}$$

A z-score of 0 is average, +2 is two standard deviations above average. NumPy does it for the entire array at once
using **broadcasting** (the single numbers `mean` and `std` are stretched to match the array).

```python
scores = np.array([72, 85, 90, 64, 78, 95, 55])
z = (scores - scores.mean()) / scores.std()
print(z.round(2))
print("mean of z:", z.mean().round(10), " std of z:", z.std().round(10))
```

**Broadcasting** also works between arrays of different shapes, as long as they line up:

```python
table = np.array([[1, 2, 3],
                  [4, 5, 6]])
col_means = table.mean(axis=0)       # shape (3,)
print(table - col_means)             # subtract each column's mean: "centering"
```
