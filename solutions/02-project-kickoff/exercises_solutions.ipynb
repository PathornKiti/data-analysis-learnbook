## Exercise 3

**Feasibility check in code.** Load `students.csv` and store in `n_rows` the number of rows and in
`columns_with_missing` the **list** of column names that contain at least one missing value
(use `df.isna().sum()`; the list may be empty).

```python
import pandas as pd

df = pd.read_csv(DATA / "students.csv")
n_rows = len(df)
missing = df.isna().sum()
columns_with_missing = list(missing[missing > 0].index)
```

```python
assert n_rows == 200
assert columns_with_missing == [], "students.csv has no missing values"
print("✅ Correct!")
```
