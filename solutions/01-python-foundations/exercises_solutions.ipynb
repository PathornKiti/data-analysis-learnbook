## Exercise 1

**Temperature conversion.** Write `c_to_f(c)` that converts Celsius to Fahrenheit using $F = \frac{9}{5}C + 32$.

```python
def c_to_f(c):
    return 9 / 5 * c + 32
```

```python
assert c_to_f(0) == 32, "0 °C should be 32 °F"
assert c_to_f(100) == 212, "100 °C should be 212 °F"
assert abs(c_to_f(-40) - (-40)) < 1e-9, "-40 is the same on both scales"
print("✅ Correct!")
```

## Exercise 2

**Your own mean.** Write `my_mean(values)` using a **loop** (not `sum`) to add up the values, then divide by how many there are.
Think about the Σ formula.

```python
def my_mean(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)
```

```python
assert my_mean([1, 2, 3]) == 2
assert my_mean([10]) == 10
assert abs(my_mean([72, 85, 90, 64, 78]) - 77.8) < 1e-9
print("✅ Correct!")
```

## Exercise 4

**Word counts.** Fill the dictionary `word_counts` so that each word in `sentence` maps to how many times it appears.
(Hint: `sentence.split()` gives a list of words; reuse the counting pattern from the lesson.)

```python
sentence = "the cat and the hat and the bat"
word_counts = {}
for word in sentence.split():
    word_counts[word] = word_counts.get(word, 0) + 1
```

```python
assert word_counts == {"the": 3, "cat": 1, "and": 2, "hat": 1, "bat": 1}, "Check your counts"
print("✅ Correct!")
```

## Exercise 5

**Variance by hand.** The *variance* measures how spread out numbers are:

$$\text{var}(x) = \frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2$$

Write `variance(values)` that returns this. You may call your own `my_mean` from Exercise 2 (or `sum(values)/len(values)`).

```python
def variance(values):
    m = sum(values) / len(values)
    return sum((v - m) ** 2 for v in values) / len(values)
```

```python
assert variance([5, 5, 5, 5]) == 0, "No spread means variance 0"
assert abs(variance([1, 2, 3, 4, 5]) - 2.0) < 1e-9
assert abs(variance([2, 4, 4, 4, 5, 5, 7, 9]) - 4.0) < 1e-9
print("✅ Correct!")
```

## Exercise 6

**BMI category.** Body-mass index is $\text{BMI} = \frac{\text{weight (kg)}}{\text{height (m)}^2}$.
Write `bmi_category(weight_kg, height_m)` returning `"underweight"` if BMI < 18.5, `"normal"` if < 25, `"overweight"` if < 30, otherwise `"obese"`.

```python
def bmi_category(weight_kg, height_m):
    bmi = weight_kg / height_m ** 2
    if bmi < 18.5:
        return "underweight"
    elif bmi < 25:
        return "normal"
    elif bmi < 30:
        return "overweight"
    return "obese"
```

```python
assert bmi_category(50, 1.80) == "underweight"
assert bmi_category(70, 1.75) == "normal"
assert bmi_category(85, 1.75) == "overweight"
assert bmi_category(110, 1.75) == "obese"
print("✅ Correct!")
```
